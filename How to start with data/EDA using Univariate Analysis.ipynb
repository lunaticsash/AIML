```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

# 1. Categorical Data

### a. Countplot

```python
sns.countplot(x='Survived', data=df, palette='Set1', hue='Sex')
```

```python
df['Survived'].value_counts().plot(kind = 'bar')
```

```python
sns.countplot(x='Pclass', data=df, palette=['green', 'blue', 'red'])
```

```python
sns.countplot(x='Embarked', data=df)
```

### b. Piechart

```python
df['Survived'].value_counts().plot(kind = 'pie', autopct='%1.1f%%')
```

```python
df['Sex'].value_counts().plot(kind = 'pie', autopct='%1.1f%%')
```

# 2. Numerical Data

### a. Histogram

```python
#for data distribution we can use histogram and kde plot
plt.hist(df['Age'], bins=20, color='skyblue', alpha=0.5)
```

### b.Displot

```python
# sns.displot(df['Age'], kde = True) 
#with probability density function
sns.displot(df['Age'], kde = True, stat = 'probability')
```

### c. BoxPlot

```python
sns.boxplot(x = 'Fare', data = df)
```

```python
sns.boxplot(x = 'Age', data = df)
#ab if u want then you can remove outliers from the data using IQR method and then plot the boxplot again to see the difference.
```

```python
df["Age"].skew() #if it is perfect 0 then it means it is normally distributed, if it is positive then it is right skewed and if it is negative then it is left skewed.
```
